```python
years = list(range(1992,2022))
import requests
from bs4 import BeautifulSoup
from io import StringIO
```

```python
url_template = "https://www.basketball-reference.com/awards/awards_{}.html"
```

```python

for year in years: 
    url = url_template.format(year) 
    data = requests.get(url)

    with open("NBA_Project/mvps/{}.html".format(year), "w+") as file:
        file.write(data.text)
```

```python
dfs= []
for year in years:
    with open ("NBA_Project/mvps/{}.html".format(year)) as file: 
        page = file.read()
    soup = BeautifulSoup(page, "html.parser")
    soup.find("tr", class_="over_header").decompose()
    mvp_tables = soup.find(id = "mvp")
    mvp = pd.read_html(StringIO(str(mvp_tables)))[0]
    mvp["Year"] = year

    dfs.append(mvp)
```

```python
mvps = pd.concat(dfs)
mvps.head()
```

```python
mvps.to_csv("mvps.csv")
```
